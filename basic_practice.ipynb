```python
# Take a number as input and print its square and cube.

number_1 = int(input("Enter a number : "))

square = number_1  ** 2
cube = number_1 ** 3

print(f"The square of the {number_1} is {square}")
print(f"The cube of the {number_1} is {cube}")
```

```python
# Take an amount and discount percentage, then calculate the final price.

amount = float(input("Enter the amount : "))
discount_percentage = float(input("Enter the discount percentage : "))

# Calculate Discount Amount 
discounted_amount = amount * (discount_percentage / 100)

print(f"Your final price would be : {round(amount - discounted_amount, 2)}\nYou've got {round(discounted_amount, 2)} discount")
```
